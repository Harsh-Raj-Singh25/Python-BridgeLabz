1. Even or Odd Checker: Check whether a given number is even or odd using an if-else statement.

```python
num=int(input("Enter a number"))
if num%2==0:
    print("Even number")
else:
    print("Odd number")

print("Even" if num % 2==0 else "odd") # Alternate
```

2. Department List Printer: Create a loop that prints all department names in the list.

```python
departments = ["HR", "Finance", "IT", "Sales"]
for dept in departments:
    print(dept)
```

3. Countdown Script: Write a script that counts down from 10 to 1 using a while loop and prints "Blast Off!" at the end.

```python
count = 10
while count > 0:
    print(count, end=' ')
    count -= 1
print("Blast Off!")
```

4. Production Target Checker: Read a list of daily production units and print the days where production was below target.

```python
production = [120, 95, 110, 80]
target = 100
for i in range(len(production)):
    if production[i]<target:
        print(f"Day {i+1} : {production[i]} (below target)")
```

5. Attendance Validator: Validate an employee's attendance percentage and categorize it as Excellent, Satisfactory, or Poor.

```python
attendance = float(input("Enter attendance percentage: "))

if attendance >= 90:
    print("Excellent")
elif 75 <= attendance < 90:
    print("Satisfactory")
else:
    print("Poor")
```

6. Odd Number Printer with Skip: Print all odd numbers between 1 and 30, skipping multiples of 5 using the continue statement.

```python
for i in range(1, 31, 2): # step of 2
    if i % 5 == 0:
        continue
    print(i, end=' ')
```

7. Order Processing Simulator: Loop through a list of orders and stop processing when a "Failed" status is encountered using break.

```python
orders = ["Pending", "Processing", "Delivered", "Failed", "Pending"]

for status in orders:
    if status == "Failed":
        print(f"Processing stopped due to {status} status.")
        break
    print(f"Processing order: {status}")
```

8. Age Validator and Counter: Accept a list of customer ages and count how many are above 18.

```python
ages = [16, 21, 18, 25, 14]
eligible_count = 0

for age in ages:
    if age > 18:
        eligible_count += 1

print(f"Number of customers above 18: {eligible_count}")
```

9. Compound Interest Calculator: Calculate compound interest over multiple years using a while loop and display the accumulated amount each year.

```python
principal = float(input("Principal: "))
rate = float(input("Rate: "))
years = int(input("Years: "))

current_year = 1

while current_year <= years:
    principal *= (1 + rate / 100)
    print(f"Year {current_year}: {principal:.2f}")
    current_year += 1
```
