1. Multi-Stage Transaction Validator: Validate incoming financial transactions using multiple layered conditions.

```python
transactions = [
    {
        "id": 101, 
        "amount": 5000, 
        "region": "EU", 
        "verified": True
    }, 
    {
        "id": 102, 
        "amount": 12000, 
        "region": "ASIA", 
        "verified": False
    },
    {
        "id": 103, 
        "amount": 10001, 
        "region": "US", 
        "verified": True
    }
]

accepted, flagged, rejected =0,0,0
for t in transactions:
    if not t["verified"]:
        rejected+=1
        print(f"transaction {t["id"]} rejected : verification failed")
    elif t["amount"]> 10000:
        flagged+=1
        print(f"transaction {t["id"]} flagged : amount concerning")
    else:
        accepted+=1
print(f"Accepted: {accepted} | Rejected: {rejected} | Flagged: {flagged}")
```

20. Expense Forecast Correction Module: Apply correction rules for deviance > 10%.

```python
forecast = [1000, 1200, 1100]
actual = [950, 1500, 1000]

for i, (f, a) in enumerate(zip(forecast, actual), 1):
    dev = abs(f - a) / f * 100
    if dev > 10:
        corr = f + (a - f) * 0.5
        print(f"Month {i} corrected: {f} -> {int(corr)} ({dev:.1f}% deviation)")
```

26. Server Uptime Log Analyzer: Identify consecutive downtime.

```python
logs = {
    "srv01": [99, 97, 94, 92], 
    "srv02": [96, 96, 97, 98], 
    "srv03": [88, 92, 91, 89]
}

for srv, up_list in logs.items():
    consecutive_low = 0
    for u in up_list:
        if u < 95:
            consecutive_low += 1
        else:
            if consecutive_low >= 2: break 
    if consecutive_low >= 4: 
        print(f"{srv}: Critical - {consecutive_low} consecutive low uptime days")
    elif consecutive_low >= 2: 
        print(f"{srv}: Warning - {consecutive_low} consecutive low uptime days")
```

27. API Call Retry and Timeout Manager: Simulate exponential retries and failures.

```python
apis = {
    "user_data": [True, False, False, True], 
    "payment": [False] * 6
}

for api, results in apis.items():
    retries = 0
    for r in results:
        if r:
            if retries > 0: 
                print(f"{api}: Succeeded after {retries} retries")
            break
        retries += 1
        if retries == 5:
            print(f"{api}: Permanently failed after 5 retries")
            break
```

29. Iterative Log Cleaner with Conditional Archiving: Categorize logs by age and size.


```python
logs = [
    {
        "file": "log1", 
        "days_old": 2, 
        "size": 5
    }, 
    {
        "file": "log2", 
        "days_old": 9, 
        "size": 0
    }, 
    {
        "file": "log3", 
        "days_old": 15, 
        "size": 3
    }
]

a, d, r = 0, 0, 0

for log in logs:
    if log["size"] == 0: 
        d += 1
    elif log["days_old"] > 7: 
        a += 1
    else: 
        r += 1
        
print(f"Archived: {a} | Deleted: {d} | Retained: {r}")
```
