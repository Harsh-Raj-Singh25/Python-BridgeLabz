Problem 1: Calculate net salary after applying tax percentage on gross salary.

```python
def netSalary(salary,tax_rate):
    return salary - (salary*tax_rate/100)
print(netSalary(100000,15))
```

Problem 2: Print a message surrounded by 40 hyphens on both sides.

```python
def print_hyphen(n): 
    return "-"*n
def format_msg(msg,n):
    return print_hyphen(n)+ msg+ print_hyphen(n)

print(format_msg("Harsh",40))
```

Problem 3: Reverse each word in a sentence individually.

```python
def reverse_words(sentence):
    # return [word[::-1] for word in sentence.split()] # for a list
    return " ".join(word[::-1] for word in sentence.split()) # for a proper sentence
# list=reverse_words("Harsh Raj Singh")
# print(list)
print(reverse_words("Harsh Raj Singh"))
```

Problem 4: Normalize names by trimming spaces and converting to lowercase.

```python
def normalize_names(nameList):
    return [name.strip().lower() for name in nameList]
print(normalize_names(["Harsh","Namrata "," Anand"]))
```

Problem 5: Load configuration value from dictionary with optional default fallback.

```python
config = {
    "host": "localhost",
    "port": 8080,
    "debug": True
}

def load_config(key, default=None):
    # .get() returns the value if key exists, otherwise returns the default
    return config.get(key, default)

# Examples
print(f"Host: {load_config('host', '127.0.0.1')}")     # Exists: 'localhost'
print(f"Timeout: {load_config('timeout', 30)}")        # Missing: returns default 30
print(f"Mode: {load_config('mode')}")                  # Missing: returns None
```

Problem 6: Measure and print execution time of a function.

```python
import time
def measure_time(func):
    start=time.time()
    func()
    end=time.time()
    return f"execution time = {(end -start)*1000} ms"
print(measure_time(lambda : sum(range(55))))
```

Problem 7: Recursively scan a directory and list all file paths. ***

```python
import os

def scan_directory(path):
    files = []

    for item in os.listdir(path):
        full_path = os.path.join(path, item)    
        if os.path.isdir(full_path):
            files.extend(scan_directory(full_path))
        else:
            files.append(full_path)
    
    return files

print(scan_directory("././"))
```

Problem 8: Invoke functions dynamically based on command name using dictionary mapping.

```python
def invoke(command_name, commands_dict):
    if command_name in commands_dict:
        return commands_dict[command_name]()
    return "Invalid command"

cmds = {
    "start": lambda: "System Starting...",
    "stop": lambda: "System Stopping...",
    "restart": lambda: "System Restarting..."
}

print(invoke("stop", cmds))
```

Problem 9: Generate Fibonacci sequence using recursion up to n terms. **

```python
def generate_fibonacci(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]
    if n == 2:
        return [0, 1]
    
    seq = generate_fibonacci(n-1)
    seq.append(seq[-1] + seq[-2])
    return seq

print(generate_fibonacci(8))
```
