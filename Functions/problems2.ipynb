Problem 2: Calculate total size of nested directories using recursion.

```python
directory_structure = {
    'folder1': {    
        'file1.txt': 100, 
        'file2.txt': 200,
        'subfolder1': { # Nested subfolder
            'file3.txt': 300
        }
    },
    'folder2': { # Another folder at the same level as folder1
        'file4.txt': 400
    }
}
def calculate_directory_size(directory):
    total_size = 0 # Initialize total size
    for item in directory.values(): # Iterate through items in the directory
        if isinstance(item, dict): # Check if the item is a subdirectory
            total_size += calculate_directory_size(item)  # Recursive call for subdirectories
        else: 
            total_size += item  # Add file size
    return total_size # Return the total size of the directory
total_size = calculate_directory_size(directory_structure) # Calculate total size of the directory
print(f'Total size of the directory: {total_size} bytes')
```

Problem 3: Generate a formatted email using name, department, and joining date.

```python
# Problem 3: Generate a formatted email using name, department, and joining date.
def generate_email(name, department, joining_date):
    email_template = f"""
    Dear {name},
Welcome to the {department} department! We are excited to have you join us on {joining_date}.
Best regards,
HR Team
    """
    return email_template.strip() # Remove leading/trailing whitespace
# Example usage
name = "Alice"
department = "Engineering"
joining_date = "2024-07-01"
email_content = generate_email(name, department, joining_date) # Generate the email content
print(email_content) # Print the generated email
```
