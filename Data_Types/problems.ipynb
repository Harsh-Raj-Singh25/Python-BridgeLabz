2. Convert user-input temperature from string to float and print Celsius/Fahrenheit equivalents.

```python
user_input=input("Enter the temperature in Celsius: ")
temp_C=float(user_input)
print(f"Celsius : {temp_C}")
temp_fahrenheit=(temp_C * 9/5) + 32
print(f"Fahrenheit : {temp_fahrenheit}")
```

4. Take a product's name, price, and discount rate, and calculate the final price

```python
name, price, discount_rate= input("Name : "), float(input("Price: ")), float(input("discount rate : "))
final_price= price - (price*discount_rate/100)
print(f"final price of the {name} : Rs.{final_price}")
```

6. Build a list of sales amounts, compute their average, and display it rounded to two decimal points.

```python
sales = [120.50, 340.75, 210.00, 450.25]
print(f"The average sales amount is Rs.{sum(sales) / len(sales):.2f}")
```

7. Dictionary of lists for monthly revenue for different branches. Compute the branch with the highest total.

```python
revenue={
    "Bihar":[50000,45000,65000],
    "UP":[48000,49750,52000],
    "MP":[55000,30000,42000],
    "Delhi":[42078,25789,48965]
}
branch=max(revenue, key=lambda b: sum(revenue[b]))
print(f"Branch with the highest total: {branch}")
```

8. Read a list of values from a CSV-like string, convert to integers, and calculate mean and maximum.

```python
csvData = "23, 45, 67, 12"
values=[int(i) for i in csvData.split(",")]
print(values)
print(f"maximum value : {max(values)} ",end="\n" f"mean value : {sum(values)/len(values)}")
```

9. Nested dictionary for departments/employees. Print all employees for a user-entered department.

```python
depts = {
    "Engineering": ["Alice", "Bob"], 
    "HR": ["David", "Eve"]
}

department = input("Enter Dept (Engineering or HR): ")
print("Employees:", ", ".join(depts.get(department, ["Department not found"])))
```
