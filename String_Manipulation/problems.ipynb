2. Format filenames by replacing spaces with underscores and converting to lowercase.

```python
def format_filename(name):
    return name.replace(" ", "_").lower()

print(format_filename("My Report FINAL .txt"))
```

3. Normalize product IDs to uppercase format with hyphens (PROD-XXX).

```python
def normalize_ids(ids):
    result = []
    for i in ids:
        i = i.replace("_", "-").upper()
        result.append(i)
    return result

print(normalize_ids(["prod-001", "PROD_002", "prod-003"]))
```

4. Extract base URL and path from a given URL string.

```python
# 4. Extract base URL and path from a given URL string.
def parse_url(url):
    if "://" in url:
        protocol, rest = url.split("://", 1)
    else:
        protocol, rest = "http", url
    if "/" in rest:
        base_url, path = rest.split("/", 1)
        path = "/" + path
    else:
        base_url, path = rest, ""
    return protocol, base_url, path
print(parse_url("https://example.com/products/item1"))
```

5. Convert raw error logs into clean, readable formatted messages.

```python
def format_errors(text):
    lines = text.split("\n")
    result = []
    
    for line in lines:
        line = line.replace("ERROR", "Error").replace("WARNING", "Warning")
        line = line.replace("at line", "line ").replace(":", " -")
        result.append(line.strip())
    
    return "\n".join(result)

text = "ERROR at line12: missing ; \nWARNING at line15: unused variable"
print(format_errors(text))
```

 Clean CSV text by removing extra symbols and normalizing spacing.

```python
def clean_csv(text):
    text = text.replace("!!", "").replace(",,", ",")
    parts = [i.strip() for i in text.split(",") if i.strip()]
    return ", ".join(parts)

print(clean_csv("John,, Doe!!, New York "))
```

7. Extract specific keywords like Python and Django from text.

```python
def extract_keywords(text):
    keywords = ["python", "django"]
    words = text.lower().split()
    
    result = []
    for k in keywords:
        if k in text.lower():
            result.append(k.capitalize())
    
    return result

print(extract_keywords("John has experience in Python, Django, and React."))
```

8. Generate dynamic emails by replacing template placeholders with data.

```python
def generate_email(template, data):
    return template.format(**data)

template = "Hello {name}, your order {order_id} is confirmed."
data = {"name": "Harsh", "order_id": "12345"}

print(generate_email(template, data))
```

9. Compress text by replacing repeated characters with character and count.

```python
def compress(text):
    result = ""
    count = 1
    
    for i in range(1, len(text)):
        if text[i] == text[i-1]:
            count += 1
        else:
            result += text[i-1] + str(count)
            count = 1
    
    result += text[-1] + str(count)
    return result

print(compress("aaabbcdddd"))
```
